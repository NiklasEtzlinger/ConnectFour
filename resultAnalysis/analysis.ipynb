```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
raw_data = pd.read_csv('ResultsConnect4.csv')
raw_data.head()
```

```python
raw_data.sum().plot(kind = 'bar')
plt.title('Total number of considered moves by MiniMax over 10 turns')
plt.xlabel('Pruning Level')
plt.ylabel('Total Considered Moves')
plt.ticklabel_format(axis='y', style='sci', scilimits=(4,4))

plt.show()
```
